### Monthly Sales

```python

df["Date"] = pd.to_datetime(df["Date"])
```

```python

df["Month"] = df["Date"].dt.month
```

```python

monthly_sales = df.groupby("Month")["Amount"].sum()
monthly_sales
```

```python
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.show()
```

### Sales by Category

```python
# First calculate total sales for each category 

category_sales = df.groupby("Category")["Amount"].sum()

category_sales
```

```python

category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

### Sales by State

```python

state_sales = df.groupby("ship-state")["Amount"].sum()

state_sales = state_sales.sort_values(ascending=False).head(10)

sns.barplot(x=state_sales.values, y=state_sales.index)

plt.title("Top 10 States by Sales")
plt.xlabel("Sales Amount")
plt.ylabel("State")
plt.show()
```

### Sales by City

```python

city_sales = df.groupby("ship-city")["Amount"].sum()

city_sales = city_sales.sort_values(ascending=False).head(10)

sns.barplot(x=city_sales.values, y=city_sales.index)

plt.title("Top 10 Cities by Sales")
plt.xlabel("Sales Amount")
plt.ylabel("City")
plt.show()
```

### Sales by Size

### Sales by Fulfilment

### Sales by Category and Size — Heatmap

```python

category_size = pd.pivot_table(
    df,
    values="Amount",
    index="Category",
    columns="Size",
    aggfunc="sum"
)

category_size
```

```python

sns.heatmap(category_size, annot=True, fmt=".0f")

plt.title("Sales by Category and Size")
plt.xlabel("Size")
plt.ylabel("Category")
plt.show()
```

### Order Count by Category and Size

```python

category_size_orders = pd.pivot_table(
    df,
    values="Order ID",
    index="Category",
    columns="Size",
    aggfunc="count"
)

category_size_orders
```

```python

sns.heatmap(category_size_orders, annot=True, fmt=".0f")

plt.title("Number of Orders by Category and Size")
plt.xlabel("Size")
plt.ylabel("Category")
plt.show()
```

### Order Count by Category and Fulfilment

```python

category_fulfilment = pd.pivot_table(
    df,
    values="Order ID",
    index="Category",
    columns="Fulfilment",
    aggfunc="count"
)

category_fulfilment
```

```python

sns.heatmap(category_fulfilment, annot=True, fmt=".0f")

plt.title("Orders by Category and Fulfilment")
plt.xlabel("Fulfilment")
plt.ylabel("Category")
plt.show()
```

### Sales by Category and Fulfilment

```python

category_fulfilment_sales = pd.pivot_table(
    df,
    values="Amount",
    index="Category",
    columns="Fulfilment",
    aggfunc="sum"
)

category_fulfilment_sales
```

```python

sns.heatmap(
    category_fulfilment_sales,
    annot=True,
    fmt=".0f"
)

plt.title("Sales by Category and Fulfilment")
plt.xlabel("Fulfilment")
plt.ylabel("Category")
plt.show()
```
